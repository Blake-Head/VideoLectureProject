# Example: Collision

```python
class collision_2d(Scene):
    def construct(self):
        obj1 = Circle(radius=0.2, color=GREEN, fill_opacity=0.3).shift(LEFT * 3 + UP * 1)
        obj2 = Circle(radius=0.3, color=RED, fill_opacity=0.3)

        obj1.generate_target()
        obj2.generate_target()

        dotted_line = DashedLine(
            start=LEFT*10,
            end=RIGHT*10,
            color=WHITE,
            dash_length=0.05,
            dashed_ratio=0.5,
        ).set_opacity(0.5)

        
        #self.add(dotted_line)
        self.play(FadeIn(obj1, obj2))

        direction = normalize(obj2.get_center() - obj1.get_center())
        obj1_velocity = Arrow(
            start=obj1.get_center() + direction * obj1.radius,
            end=obj2.get_center() - direction * obj2.radius*(8),
            color=GREEN,
            buff=0,
            stroke_width=5
        )


        dotted_line2 = DashedLine(
            start=obj1.get_center(),
            end=obj2.get_center(),
            color=GREEN,
            dash_length=0.05,
            dashed_ratio=0.5,
        ).set_opacity(0.5)

        #self.play(Create(dotted_line2))

        self.play(obj1.animate.move_to(obj2.get_edge_center(UP)+UP*0.2), run_time=2, rate_func=linear)

        self.play(obj1.animate.shift(RIGHT * 1.5 + UP * 0.75), obj2.animate.shift(RIGHT * 0.5 + DOWN * 1.5), run_time=2, rate_func=linear)

        dotted_line3 = DashedLine(
            start=ORIGIN,
            end=obj1.get_center(),
            color=GREEN,
            dash_length=0.05,
            dashed_ratio=0.5,
        ).set_opacity(0.5)

        dotted_line4 = DashedLine(
            start=ORIGIN,
            end=obj2.get_center(),
            color=RED,
            dash_length=0.05,
            dashed_ratio=0.5,
        ).set_opacity(0.5)

        #self.play(Create(dotted_line3), Create(dotted_line4))


        self.play(Create(dotted_line), FadeIn(obj1_velocity, obj1.copy().move_to(LEFT*3 + UP*1).set_opacity(0.1), obj2.copy().move_to(ORIGIN).set_opacity(0.1)))
        self.play(Create(dotted_line2))

        obj1_velocity_final = Arrow(
            start=ORIGIN,
            end=obj1.get_center(),
            color=GREEN,
            buff=0,
            stroke_width=3
        ).shift(RIGHT * 1.3 + UP * 1.1).scale(0.50)

        obj2_velocity_final = Arrow(
            start=ORIGIN,
            end=obj2.get_center(),
            color=RED,
            buff=0,
            stroke_width=3
        ).shift(RIGHT * 0.45 + DOWN * 1.37).scale(0.50)

        self.play(Create(dotted_line3), Create(dotted_line4), GrowArrow(obj1_velocity_final), GrowArrow(obj2_velocity_final), run_time=1, rate_func=linear)

        self.wait(1)

        self.play(
            *[FadeOut(mob)for mob in self.mobjects]
        )


# Load the manim extension
%load_ext manim

# Render the main scene
%manim -qk -v WARNING collision_2d
```

# Example: Freebody Ramp

```python
class ramp(Scene):
    def construct(self):
            ramp = Polygon(ORIGIN, LEFT*4, LEFT*4 + UP*2, color=BLUE, fill_opacity=0.1).scale(2).shift(DOWN*1 + RIGHT*2)
            obj = Circle(radius=.5, color=GREEN, fill_opacity=0.3).shift(UP*0.6)

            angle_indicator = Arc(
                start_angle=PI/2,
                angle=PI/2,
                radius=0.5,
                color=WHITE
            ).shift(DOWN*2+RIGHT*3)

            ramp_angle = MathTex(r"\theta", font_size=45).next_to(angle_indicator, LEFT).shift(UP*0.2)

            self.play(DrawBorderThenFill(ramp))
            self.play(FadeIn(obj))

            self.play(Create(angle_indicator), Write(ramp_angle))

            center_dot = Dot(color=WHITE).move_to(obj.get_center())


            grav_force = Arrow(
                start=obj.get_center(),
                end=obj.get_center() + DOWN*2.5,
                color=RED,
                buff=0,
                stroke_width=3
            )

            grav_label = MathTex("mg", font_size=45, color=RED).next_to(grav_force, DOWN)

            verts = ramp.get_vertices()

            # Sloped ramp face (triangle points are [0, 1, 2])
            a, b = verts[2], verts[0]

            tangent = normalize(b - a)                      # along the slope
            n1 = np.array([-tangent[1], tangent[0], 0.0])  # 90 deg rotation
            n2 = -n1

            # Pick outward normal consistently
            mid = 0.5 * (a + b)
            center = ramp.get_center()
            normal = n1 if np.dot(n1, mid - center) > 0 else n2

            normal_force = Arrow(
                start=obj.get_center(),
                end=mid - normal * 2.5,
                color=BLUE,
                buff=0,
                stroke_width=3
            )

            normal_label = MathTex("N", font_size=45, color=BLUE).next_to(normal_force, UP).shift(RIGHT*1)

            fric_force = Arrow(
                start=obj.get_center(),
                end=obj.get_center() + LEFT*1.5 + UP*0.75,
                color=YELLOW,
                buff=0,
                stroke_width=3
            )   

            fric_label = MathTex("f", font_size=45, color=YELLOW).next_to(fric_force, LEFT).shift(UP*0.5)

            self.play(FadeIn(center_dot), GrowArrow(grav_force), Write(grav_label), GrowArrow(normal_force), Write(normal_label), GrowArrow(fric_force), Write(fric_label))
            self.wait(1)

            self.wait(1)

            self.play(
                *[FadeOut(mob)for mob in self.mobjects]
            )



# Load the manim extension
%load_ext manim

# Render the main scene
%manim -qk -v WARNING ramp
```

# Example: Electric Fields

```python
class efield(Scene):
    def construct(self):
        charge1 = mp.Charge(magnitude=2, point=np.array([-2, 0, 0]), add_glow = True)
        charge2 = mp.Charge(magnitude=-2, point=np.array([2, 0, 0]), add_glow = True)

        field1 = mp.ElectricField(charge1, color=RED)
        field2 = mp.ElectricField(charge2, color=BLUE)

        field3 = mp.ElectricField(charge1, charge2)

        #self.add(charge1, charge2)
        self.add(charge1, charge2, field1, field2)
        self.wait(1)
        self.play(FadeOut(field1), FadeOut(field2), FadeIn(field3))
        self.wait(1)
        self.play(FadeOut(field3), FadeIn(field1), FadeIn(field2))
        self.wait(1)


# Load the manim extension
%load_ext manim

# Render the main scene
%manim -qk -v WARNING efield
```

# Intro Description

```python
class s2p1(Scene):
    def construct(self):
        Text.set_default(font="Gill Sans")
        number_plane = NumberPlane(
            axis_config={
                "stroke_color": WHITE,
                "stroke_width": 0,
            },
            background_line_style={
                "stroke_color": BLUE_D,
                "stroke_width": 4,
                "stroke_opacity": 0.2,

            }
        )# Axis labels
        x_label = MathTex("x").next_to(number_plane.c2p(6, 0), RIGHT*1.5+DOWN*0.7)
        y_label = MathTex("y").next_to(number_plane.c2p(0, 3), UP+RIGHT*0.7+UP*0.2)


        # Starting Setup

        dot = Dot(color=PURE_GREEN, radius=0.1).move_to(number_plane.c2p(0, 0))

        self.play(GrowFromCenter(dot))
        self.wait(1)
        self.play(Create(number_plane))
        self.wait(1)

        self.play(dot.animate.move_to(number_plane.c2p(3, 0)), run_time=2, rate_func=linear)
        self.wait(1)

        start_arrow = Arrow(
            start=number_plane.c2p(0, -2),
            end=number_plane.c2p(0, 0),
            color=YELLOW,
            buff=0,
            stroke_width=3
        )
        here_text = Text("Here", font_size=30, color=YELLOW).next_to(start_arrow, DOWN)
        self.play(GrowArrow(start_arrow))
        self.play(Write(here_text))
        self.wait(1)

        end_arrow = Arrow(
            start=number_plane.c2p(3, -2),
            end=number_plane.c2p(3, 0),
            color=YELLOW,
            buff=0,
            stroke_width=3
        )
        there_text = Text("There", font_size=30, color=YELLOW).next_to(end_arrow, DOWN)
        self.play(GrowArrow(end_arrow))
        self.play(Write(there_text))
        self.wait(1)


        # Word Scroll

        prefix = Text("3", font_size=35)
        word_list = ["units", "meters", "feet", "miles", "yards", "football fields", "light-years", "units"]
        words = VGroup(*[
            Text(word, font_size=35)
            for word in word_list
        ]).arrange(DOWN, buff=0.0, aligned_edge=LEFT)

        words.move_to(ORIGIN+RIGHT*2.6)
        prefix.next_to(words[0], LEFT, buff=0.2)
        prefix.align_to(words[0], DOWN)

        focus_y = words[0].get_y()
        fade_distance = 0.35

        def update_word_opacity(mob):
            distance = abs(mob.get_y() - focus_y)
            opacity = np.clip(1 - distance / fade_distance, 0.0, 1.0)
            mob.set_opacity(opacity)

        for word in words:
            word.add_updater(update_word_opacity)
            update_word_opacity(word)

        self.play(FadeIn(prefix), FadeIn(words[0], shift=DOWN * 0.15))
        self.wait(0.5)

        for i in range(1, len(words)):
            delta_y = focus_y - words[i].get_y()
            self.play(
                words.animate.shift(UP * delta_y),
                run_time=0.50,
                rate_func=linear,
            )

        self.wait(1)

        self.play(FadeOut(here_text), FadeOut(there_text), FadeOut(start_arrow), FadeOut(end_arrow))    
        self.wait(1)

        # Make new number plane just to fade in x- and y-axes

        number_plane1 = NumberPlane(
            axis_config={
                "stroke_color": WHITE,
                "stroke_width": 2,
            },
            background_line_style={
                "stroke_color": BLUE_D,
                "stroke_width": 4,
                "stroke_opacity": 0.2,

            }
        )

        central_dot = Dot(color=WHITE, radius=0.07).move_to(number_plane.c2p(0, 0))

        self.play(FadeIn(number_plane1), FadeOut(number_plane), GrowFromCenter(central_dot))
        self.wait(1)
        self.play(Write(x_label), Write(y_label))
        self.wait(1)

        new_prefix = Text("in the x-direction", font_size=35)
        new_prefix.next_to(words[7], RIGHT, buff=0.2)
        new_prefix.align_to(words[7], DOWN)
        self.play(Write(new_prefix))
        self.wait(1)

        right_arrow = Arrow(
            start=ORIGIN,
            end=number_plane.c2p(1, 0),
            color=BLUE,
            buff=0,
            stroke_width=3
        )

        left_arrow = Arrow(
            start=ORIGIN,
            end=number_plane.c2p(-1, 0),
            color=BLUE,
            buff=0,
            stroke_width=3
        )

        self.play(GrowArrow(right_arrow), GrowArrow(left_arrow))
        self.wait(1)

        other_dot = Dot(color=PURE_GREEN, radius=0.1).move_to(number_plane.c2p(-3, 0)).set_opacity(0.3)
        self.play(FadeIn(other_dot))
        self.wait(1)

        new_prefix = Text("in the positivex-direction", font_size=35)
        new_prefix.next_to(words[7], RIGHT, buff=0.2)
        new_prefix.align_to(words[7], DOWN)
        self.play(Write(new_prefix))
        self.wait(1)



# Load the manim extension
%load_ext manim

# Render the main scene
%manim -ql -v WARNING s2p1
```
